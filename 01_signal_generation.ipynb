```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

print("RF signal generation module initialized!")
```

```python
def generate_bpsk(n_symbols):

    bits = np.random.randint(0, 2, n_symbols)

    symbols = 2 * bits - 1

    return symbols.astype(complex)


bpsk = generate_bpsk(1000)

print("BPSK generated:", len(bpsk))
```

```python
def generate_qpsk(n_symbols):

    i = np.random.choice([-1, 1], n_symbols)
    q = np.random.choice([-1, 1], n_symbols)

    symbols = (i + 1j*q) / np.sqrt(2)

    return symbols


qpsk = generate_qpsk(1000)

print("QPSK generated:", len(qpsk))
```

```python
def generate_16qam(n_symbols):

    levels = np.array([-3, -1, 1, 3])

    i = np.random.choice(levels, n_symbols)
    q = np.random.choice(levels, n_symbols)

    symbols = (i + 1j*q) / np.sqrt(10)

    return symbols


qam16 = generate_16qam(1000)

print("16-QAM generated:", len(qam16))
```

```python
def plot_constellation(symbols, title):

    plt.figure(figsize=(5, 5))

    plt.scatter(
        np.real(symbols),
        np.imag(symbols),
        s=10,
        alpha=0.5
    )

    plt.title(title)
    plt.xlabel("In-phase (I)")
    plt.ylabel("Quadrature (Q)")

    plt.grid(True)
    plt.axis("equal")
    plt.show()


plot_constellation(bpsk, "BPSK Constellation")
plot_constellation(qpsk, "QPSK Constellation")
plot_constellation(qam16, "16-QAM Constellation")
```

```python
print("BPSK power:", np.mean(np.abs(bpsk)**2))
print("QPSK power:", np.mean(np.abs(qpsk)**2))
print("16-QAM power:", np.mean(np.abs(qam16)**2))
```
